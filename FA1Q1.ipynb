```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import pandas as pd

df = pd.read_csv("/content/export.csv")
df.head()
```

```python
df.columns
```

```python
print(df.info())
print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Remove rows with missing values
df = df.dropna()

# Convert sales columns to numeric
df["RETAIL SALES"] = pd.to_numeric(df["RETAIL SALES"], errors="coerce")
df["RETAIL TRANSFERS"] = pd.to_numeric(df["RETAIL TRANSFERS"], errors="coerce")
df["WAREHOUSE SALES"] = pd.to_numeric(df["WAREHOUSE SALES"], errors="coerce")

# Remove any new missing values created after conversion
df = df.dropna()

print(df.info())
```

```python
monthly_sales = df.groupby("MONTH")["RETAIL SALES"].sum()

monthly_sales.plot(kind="bar", figsize=(8,5))

plt.title("Monthly Retail Sales")
plt.xlabel("Month")
plt.ylabel("Total Retail Sales")
plt.show()
```
